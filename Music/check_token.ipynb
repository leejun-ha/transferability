```python
import transformers
```

```python
def count_unused_tokens(model_name):
    # Load the tokenizer for the specified model
    tokenizer = transformers.AutoTokenizer.from_pretrained(model_name)
    
    # Access the vocabulary
    vocab = tokenizer.get_vocab()
    
    # Count unused tokens
    unused_tokens_count = sum(1 for token in vocab if token.startswith("[unused"))
    
    return unused_tokens_count

# Example usage
models = ["bert-base-uncased", "bert-base-chinese", "bert-base-multilingual-uncased", "bert-base-multilingual-cased", 
          "bert-base-german-cased", "wietsedv/bert-base-dutch-cased", "neuralmind/bert-base-portuguese-cased",	
          "aubmindlab/bert-base-arabert", "tohoku-nlp/bert-base-japanese", "kykim/bert-kor-base", "bertin-project/bertin-roberta-base-spanish", 
"microsoft/codebert-base", "microsoft/codebert-base-mlm", "neulab/codebert-javascript", "neulab/codebert-java", "neulab/codebert-python", "neulab/codebert-c"
]
for model in models:
    count = count_unused_tokens(model)
    print(f"Number of unused tokens in {model}: {count}")
```
